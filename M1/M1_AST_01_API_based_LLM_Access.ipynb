####  **Making an API call for OpenAI Model**

This cell demonstrates how to make a chat completion API call using the initialized Groq client.

- A `prompt` variable is defined containing the instructions for the LLM to generate an essay.

- `client.chat.completions.create()` is called with the following parameters:
    - `model`: Specifies the model to use (`openai/gpt-oss-120b` in this case).
    - `messages`: A list of message objects representing the conversation history. In this case, it contains a single message with the role "user" and the content being the `prompt`.

- The response from the API is stored in the `response` variable.

- `print(response.choices[0].message.content)` extracts the generated text content from the response and prints it to the console.

````python
# Define a multi-line string prompt using triple quotes
# This prompt instructs the AI to write an essay on AI and climate change

prompt = """Write a detailed but concise essay (400–500 words) on the topic:
“The role of artificial intelligence in addressing global climate change.”

Your essay should include:

1. A short introduction (2–3 sentences)
2. 2–3 key arguments supported by real-world examples
3. A brief conclusion summarizing your viewpoint

Make sure the content is:
Clear and logically structured
Factually correct
Engaging and easy to follow
"""
````

````python
# Send the prompt to the Groq Chat Completion API
# This creates a request to generate a response from the LLM

response = groq_client.chat.completions.create(
    model="openai/gpt-oss-120b",                          # Specify the LLM to be used
    messages=[                                            # Provide the conversation as a list of messages
        {"role": "user", "content": prompt}               # Here, we send only one user message containing the prompt
    ]
)
````

````python
# Print the LLM-generated response
# `response.choices[0]` selects the first generated answer
# `.message.content` extracts the actual text output

print(response.choices[0].message.content)
````

Similarly, we can access LLMs via API from other model providers as well such as Google Gemini API, etc.

####  **Making an API call for Llama Model**

````python
# Send the prompt to the Groq Chat Completion API
# This creates a request to generate a response from the LLM

response = groq_client.chat.completions.create(
    model="meta-llama/llama-4-scout-17b-16e-instruct",    # Specify the LLM to be used
    messages=[                                            # Provide the conversation as a list of messages
        {"role": "user", "content": prompt}               # Here, we send only one user message containing the prompt
    ]
)
````

````python
print(response.choices[0].message.content)
````

## **Google Gemini API**

- Go to https://ai.google.dev/gemini-api/docs

- Click on **Get a Gemini API key**

- Copy an existing key or Create a new one

- Save the key in Google Colab Secrets with name `GEMINI_API_KEY`

- Use the below code to make an API call using Google Gemini API

````python
# Save the key in Colab's Secrets then load from there
import os
from google.colab import userdata
os.environ['GEMINI_API_KEY'] = userdata.get('GEMINI_API_KEY')
````

````python
# Initialize Client
from google import genai

gemini_client = genai.Client()
````

````python
# Provide your prompt
prompt = "Which was built first - Taj Mahal, or Eiffel Tower"

# Make API call and get response
gemini_response = gemini_client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt,
)

print(gemini_response.text)
````
