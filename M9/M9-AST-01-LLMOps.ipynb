### **What is LLMOps?**

**LLMOps** is the set of practices used to manage the complete lifecycle of LLM-powered applications. It includes model access, prompt management, RAG monitoring, evaluation, guardrails, tracing, deployment, cost tracking, and continuous improvement.

LLMOps helps move an LLM application from a simple prototype to a reliable production-ready product.

---

### **LLMOps Architecture**

A team that thinks of an LLM app as "a model and a prompt" hits the same walls everyone else has hit. The same product, viewed as a stack of six layers, surfaces the questions that actually need answers:

- where does the model come from,
- how do we reach it,
- what turns it into behavior,
- what makes that behavior reliable,
- what does a user touch, and
- how does what users do feed back into the system.

Each layer is a decision space with its own trade-offs.

### **1. Upstream AI**

The Upstream AI layer contains the core AI capabilities used by the application.

This includes:

* **Foundation models**
    - Generate text, reason, and follow instructions
    - Examples: Claude, GPT-4o, Gemini, Llama, Mistral, DeepSeek, Qwen
* **Embedding models**
    - Convert text → vectors for semantic search & RAG
    - Examples: text-embedding-3, BGE, Cohere Embed, e5, sentence-transformers/all-MiniLM-L6-v2
* **Rerankers**
    - Score retrieved passages by relevance — improves RAG precision
    - Examples: Cohere Rerank, ColBERT, BGE reranker

Who builds these models: Anthropic, OpenAI, Google, Meta, Mistral, DeepSeek, Alibaba

---

### **2. Model Access**

The **Model Access** layer defines how the application connects to LLMs.

The same model can sit behind a hosted API, a local runtime, or a self-hosted gateway. Same capability; different cost, latency, and privacy profile.

- **Hosted APIs**
    - Anthropic, OpenAI, Bedrock, Azure OpenAI, Google AI Studio

- **Local runtimes**
    - Ollama, LM Studio, llama.cpp, vLLM

- **Gateways**
    - LiteLLM, Portkey, TGI

This layer is important because different teams may use different models for cost, privacy, latency, or compliance reasons.

---


### **3. Application Tooling**

The **Application Tooling** layer contains the components used to build the LLM application logic.

It turns raw model capability into behavior aligned with your product. Prompts, RAG, workflows, and agents are the levers you control.

- **Prompting**
    - System messages, Few-shot, Structured output, Tool calling

- **RAG**
    - Chunking, Chroma, Pinecone, Retrieval, Reranking

- **Orchestration**
    - LangGraph
    - LlamaIndex Workflows
    - Agent frameworks

This layer defines how the LLM is actually used inside the application.

---


### **4. LLMOps Layer**

The **LLMOps layer** is the operational control layer for LLM applications.

The discipline of operating an LLM application reliably — with evidence, safety, traceability, and safe deployment.

---

- **Evaluation**
    - Checks whether the LLM application is producing good, safe, and useful responses
    - Implementation Examples: Golden datasets, LLM-as-judge, RAGAS, DeepEval
    - Evaluation can include:

        | Evaluation Type | Purpose                                                    |
        | --------------- | ---------------------------------------------------------- |
        | Correctness     | Checks whether the answer is factually correct             |
        | Relevance       | Checks whether the answer addresses the user query         |
        | Faithfulness    | Checks whether the answer is grounded in retrieved context |
        | Safety          | Checks whether the response avoids harmful content         |
        | Latency         | Measures response time                                     |
        | Cost            | Measures token and API usage                               |
        | Tool Accuracy   | Checks whether the agent selected the correct tool         |
        | RAG Quality     | Checks whether retrieved documents are useful              |

---

- **Guardrails**
    - Safety checks added before and after the LLM call
    - Implementation Examples: llm-guard, Guardrails AI, NeMo Guardrails

        | Guardrail Type              | Purpose                                             |
        | --------------------------- | --------------------------------------------------- |
        | Input Guardrails            | Validate user input before sending it to the LLM    |
        | Output Guardrails           | Validate model output before showing it to the user |
        | Topic Guardrails            | Block restricted topics                             |
        | PII Guardrails              | Detect sensitive personal information               |
        | Toxicity Guardrails         | Detect harmful or offensive content                 |
        | Prompt Injection Guardrails | Detect attempts to override system instructions     |
        | Hallucination Guardrails    | Check whether the answer is grounded in context     |

---

- **Tracing & Observability**
    - Tracing records the complete execution path of an LLM request.
    - A trace may include:
        - User input, Prompt template, Retrieved documents, LLM response, Tool calls, Agent steps, Token usage, Latency, Errors, Final output
    - Tracing is very useful for debugging RAG pipelines, agents, and multi-step workflows.
    - Example tools: Langfuse, LangSmith, Arize Phoenix, Helicone, OpenTelemetry-LLM

---

- **Monitoring**
    - Monitoring tracks the health and performance of an LLM application in production.
    - Important metrics include:

        | Metric             | Purpose                                |
        | ------------------ | -------------------------------------- |
        | Total Requests     | Measures usage volume                  |
        | Latency            | Tracks response speed                  |
        | Error Rate         | Detects failures                       |
        | Token Usage        | Tracks input and output tokens         |
        | Cost per Request   | Helps control spending                 |
        | User Feedback      | Captures satisfaction                  |
        | Hallucination Rate | Tracks unsupported answers             |
        | Safety Violations  | Tracks unsafe responses                |
        | Retrieval Quality  | Tracks RAG performance                 |
        | Model Usage        | Shows which models are being used most |

---

- **CI/CD for LLM Applications**
    - CI/CD in LLMOps means automatically testing and deploying changes to prompts, chains, tools, RAG pipelines, and application code.
    - In LLMOps, CI/CD may include:
        * Prompt regression testing
        * Evaluation dataset testing
        * RAG retrieval testing
        * Guardrail testing
        * API testing
        * Docker image build
        * Deployment to cloud or Hugging Face Spaces
        * Rollback to previous prompt or model version

    - Example:

        ```text
        Developer updates a prompt
                ↓
        Evaluation test cases run automatically
                ↓
        Guardrail checks are executed
                ↓
        Only if tests pass, the updated prompt is deployed
        ```

---

- **Governance**
    - Ensures that LLM applications follow organizational, legal, and security requirements.
    - Includes:
        * Access control
        * Audit logs
        * Model approval process
        * Prompt versioning
        * Data privacy
        * Cost budgets
        * Human review
        * Risk classification
        * Compliance checks

---


### **5. Application / Product Layer**

This layer is where users interact with the LLM system.

Examples:

* Jupyter notebooks
* Streamlit apps
* FastAPI APIs
* Dashboards
* Chatbots
* Internal business tools
* n8n workflows
* Gradio interfaces
* React applications

This layer consumes the LLM pipeline and exposes it to end users.

Example:

```text
User asks a question in Streamlit
        ↓
FastAPI receives the request
        ↓
RAG pipeline retrieves documents
        ↓
LLM generates answer
        ↓
Guardrails validate response
        ↓
Final answer is shown to the user
```

---

### **6. Users and Feedback**

The final layer captures how real users interact with the application.

Useful feedback signals include:

* User ratings
* Thumbs up/down
* Reported incorrect answers
* Logs
* Incidents
* Failed queries
* Escalations to human support
* Repeated unanswered questions
* High-cost queries
* Slow responses

This feedback is used to improve the system continuously.

# **1. LiteLLM**

<center>
<img src='https://docs.litellm.ai/assets/ideal-img/hero.cc9658c.1280.png' width=700px>
</center>


**LiteLLM** is an open-source library that gives you a single, unified interface to call 100+ LLMs — OpenAI, Anthropic, Vertex AI, Bedrock, and more — using the OpenAI format.

- Call any provider using the same `completion()` interface — no re-learning the API for each one
- Consistent output format regardless of which provider or model you use
- Built-in retry / fallback logic across multiple deployments via the [Router](https://docs.litellm.ai/docs/routing)


LiteLLM helps teams avoid writing separate integration code for every model provider. Instead, the application can call different providers through one common interface.

````python
############ If Using OpenAI Models ##############

# 3 Reasoning and 2 Non-Reasoning Models

model_list = [
    {
        "model_name": "reasoning",
        "litellm_params": {              # params for litellm completion/embedding call
            "model": "openai/gpt-5",
            "api_key": os.getenv("OPENAI_API_KEY"),
        }
    },
    {
        "model_name": "reasoning",
        "litellm_params": {
            "model": "openai/o3-mini",
            "api_key": os.getenv("OPENAI_API_KEY"),
        }
    },
    {
        "model_name": "reasoning",
        "litellm_params": {
            "model": "openai/o4-mini",
            "api_key": os.getenv("OPENAI_API_KEY"),
        }
    },
    {
        "model_name": "non-reasoning",
        "litellm_params": {
            "model": "openai/gpt-4o-mini",
            "api_key": os.getenv("OPENAI_API_KEY"),
        }
    },
    {
        "model_name": "non-reasoning",
        "litellm_params": {
            "model": "openai/gpt-4.1",
            "api_key": os.getenv("OPENAI_API_KEY"),
        }
    }
]
````

````python
# Initialize the LiteLLM Router
# The router uses `model_list` to map logical model aliases
# (e.g., "non-reasoning", "reasoning") to actual provider deployments

router = Router(model_list=model_list)
````

````python
# `router.acompletion()` is a drop-in replacement for `openai.ChatCompletion.create`
# Requests using model="non-reasoning" will be routed to a fast, low-cost deployment (e.g., gpt-4o-mini or gpt-4.1)

response = await router.acompletion(
    model="non-reasoning",
	messages=[{"role": "user", "content": "How many 'r' are there in the word Straberry?"}]
    )

print(response.choices[0].message.content)
````

````python
# Requests using model="reasoning" will be routed to a stronger, reasoning-capable deployment (e.g.,gpt-5, o3-mini, or o4-mini)

response = await router.acompletion(
    model="reasoning",
	messages=[{"role": "user", "content": "How many 'r' are there in the word Straberry?"}]
    )

print(response.choices[0].message.content)
````

### **LiteLLM - Fallbacks**

````python
from litellm import Router

router = Router(
    model_list=model_list,
    fallbacks=[{"non-reasoning": ["reasoning"]}]            # <--- KEY CHANGE
    )
````

````python
response = router.completion(
	model="non-reasoning",
	messages=[{"role": "user", "content": "How many 'r' are there in the word Straberry?"}],
	mock_testing_fallbacks=False,             # <--- To test the fallback, change this value to True
)

print(response.choices[0].message.content)
````

````python
response = router.completion(
	model="non-reasoning",
	messages=[{"role": "user", "content": "How many 'r' are there in the word Straberry?"}],
	mock_testing_fallbacks=True,
)

print(response.choices[0].message.content)
````
