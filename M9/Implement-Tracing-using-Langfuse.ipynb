# **Langfuse**

Langfuse is an open-source LLM engineering platform that helps teams collaboratively debug, analyze, and iterate on their LLM applications.

It offers ***Traces***, ***evals***, ***prompt management*** and ***metrics*** to debug and improve your LLM application.

### **1. Get Started with Tracing**

https://langfuse.com/docs/observability/get-started

To get started with Langfuse, you can either

- [Create Langfuse account](https://cloud.langfuse.com/auth/sign-up) (free tier),

    - Sign Up in Langfuse and Login
    - Create a New Organization in Langfuse.
    - Create a New Project.
    - Once the Project is ready, go to Settings (on the Left Panel) >> API Keys >> Create new API keys >> Provide a name to the Key >> Copy the Secret Key, Public Key, Base URL and paste in the code cells below.

    OR

- [Self-host Langfuse](https://langfuse.com/self-hosting)

    1. Start a new GitHub Codespace, then change its Machine Type to 4-core, and restart the Codespace.

    1. Requirements: *git*, *docker*, and *docker compose*

    2. Clone Langfuse Repository

        `git clone https://github.com/langfuse/langfuse.git`

    3. Create a `.env` file and add your OpenAI API Key

        `OPENAI_API_KEY=ADD-YOUR-KEY-HERE`

    4. Load the Environment variables (eg. OPENAI_API_KEY)

        `source .env`


    5. Start the application

        `cd langfuse`

        `docker compose up`

        Watch the docker containers being started and the logs flowing in. After about 2-3 minutes, the ***langfuse-web-1*** container should log **“Ready”**.

    6. Open http://localhost:3000 in your browser to access the Langfuse UI.

        (For GitHub Codespace it will be the ***Forwarded Address***. Make sure to ***make the Port Public***.)

        > Next, Sign Up in Langfuse. Once Signed Up, it will take you to the localhost URL, at this point visit the Forwarded Address of GitHub Codespaces again to access Langfuse.

        > Create a New Organization in Langfuse.

        > Create a New Project.

        > Once the Project is ready, go to *Settings (on the Left Panel) >> API Keys >> Create new API keys* >> Provide a name to the Key >> Copy the Secret Key, Public Key, Base URL and paste in the code cell below.

    6. To Stop the application later

        `docker compose down`

### **3. LangChain (Python)**

#### **3.1 Implement Tracing for a RAG application**

1. Start a new GitHub Codespace (machine type 2-core is also fine).

2. Download the Zip file containing the application code from LMS.

3. Upload the below files/folders on your Codespace via drag-and-drop.

    - `requirements.txt`
    - `env.txt`
    - `rag/`
    - `agent/`

4. Rename the `env.txt` file to `.env` and add your API keys for respective tools

4. Start a new Terminal and create a Virtual Environment

    `>> python -m venv venv`

5. Activate the Virtual Environment

    `>> source venv/bin/activate`

6. Review the libraries listed in `requirements.txt` file and Install them

    `>> pip install -r requirements.txt`

7. Run the RAG application

    - Move into rag/ folder:
        `>> cd rag`
    - First, create the knowledge base: `>> python create_kbase.py`
    - Then, query the knowledge base: `>> python query_kbase.py`


8. Check the traces being logged in Langfuse

    - Review the input provied to the rag chain.
    - Review the documents being retreived for the given query.
    - Review the final response from the rag chain.
    - Review the latency and cost.



#### **3.2 Implement Tracing for an Agent application**

1. Run the Agent application

    - Move into agent/ folder:
        `>> cd ../agent`
    - Query the agent: `>> python query_agent.py`


2. Check the traces being logged in Langfuse

    - Review the input provied to the agent.
    - Review which tools being called, what's the input and output from them.
    - Review the final response from the agent.
    - Review the latency and cost.
