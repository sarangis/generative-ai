## Information

**What is Fine-tuning?**

Fine-tuning is the process of taking a pre-trained Large Language Model (LLM) and adapting it to a specific task, domain, or dataset.

Instead of training a model from scratch—which requires enormous amounts of data and compute—fine-tuning leverages the knowledge the model has already learned from vast amounts of general text and specializes it for narrower use cases.

---

**Key Points:**

- **Base model:** A large model (e.g., GPT, Llama, Falcon, Mistral) already trained on broad, general-purpose text.

- **Fine-tuning process:** Continue training the model using smaller, domain- or task-specific datasets (e.g., legal documents, medical reports, customer service conversations).

- **Goal:** Improve performance on specialized tasks (classification, summarization, sentiment analysis, Q&A, chatbots, etc.) while reducing errors and irrelevant outputs.

---

**Example:**

- A pre-trained model like **Llama-2** can handle general-purpose text generation.

- By fine-tuning it on a **set of banking customer queries**, it becomes highly effective at answering **finance-specific customer support questions** (e.g., loan eligibility, account issues, compliance policies).

---

**Why Fine-tuning Matters:**

- **Domain adaptation** – Makes the model useful for specialized fields like healthcare, law, or finance.

- **Performance boost** – Increases accuracy for specific downstream tasks.

- **Efficiency** – Requires far less data and compute compared to training from scratch.

- **Customization** – Aligns the model’s outputs with organizational goals and tone.

---

**Why Fine-tune when we already have powerful pre-trained LLMs?**

Pre-trained LLMs are excellent generalists, but they:

- **Lack domain expertise** (e.g., medical, legal, financial knowledge).

- **Struggle with brand tone** (e.g., formal HR announcements vs casual marketing posts).

- **May be inefficient** for specific tasks, requiring more tokens and higher cost.

- **Are not updated for your organization’s private data** (internal FAQs, policies, product documentation).

Fine-tuning solves these by making the model:

- More **domain-adapted**.

- More **consistent** in tone and style.

- More **cost-efficient** (especially with smaller or compressed models).

---

**Distillation & SLMs (Small Language Models)**

Instead of always using large LLMs (with billions of parameters), organizations sometimes **compress knowledge** into smaller models:

- **Distillation:** Training a smaller model (the student) to mimic the behavior and outputs of a larger, more capable model (the teacher).

- **SLM (Small Language Model):** These are the compact models optimized for:

    - **Low latency** (faster responses).

    - **Lower cost** (cheaper inference, less hardware).

    - **Specialization** (only trained for a specific task, not general knowledge).

- **When useful:** Edge devices, real-time systems, enterprise tools where speed and cost matter more than broad knowledge.

---

**Fine-tuning an LLM vs Fine-tuning an SLM**

- **LLM Fine-tuning:**

    - Expensive but powerful.

    - Best when you need **broad + deep** understanding.

    - Useful for multi-task agents, reasoning-heavy jobs.

- **SLM Fine-tuning:**

    - Cheaper, faster, and task-specific.

    - Best when you only need **specialized performance** (e.g., HR email generation, customer support FAQs).

    - Easier to deploy in enterprise or edge devices.

---

**Different Strategies to Adapt or Customize LLMs without Fine-tuning**

You can often get strong results **without changing the model weights**:

- **Prompt Engineering:** Designing better prompts with clear instructions, examples, or role assignments.

- **Parameter Tuning:** Adjusting generation parameters like *temperature*, *top-p*, *max tokens* for more creativity, focus, or brevity.

- **Context Engineering / RAG (Retrieval-Augmented Generation):** Supplying relevant external knowledge (documents, FAQs, databases) at runtime so the model doesn’t need retraining.

These approaches are **low-cost and fast** — first choice before considering fine-tuning.

---

**When not to go for Fine-tuning**

You should **avoid Fine-tuning** when the requirement can be solved using simpler or less expensive techniques.

Fine-tuning is usually not the first choice in many scenarios:

* **When prompt engineering is sufficient** – If better instructions, role prompting, or few-shot examples already produce good results.
* **When knowledge changes frequently** – Fine-tuned models do not automatically learn new information; techniques like RAG are better for dynamic content.
* **When the goal is adding external knowledge** – Fine-tuning changes behavior and style more than it serves as a knowledge database.
* **When limited training data is available** – Small or poor-quality datasets can cause overfitting and degrade model performance.
* **When cost and infrastructure are constraints** – Fine-tuning requires computational resources, training time, and maintenance effort.
* **When model behavior should remain general-purpose** – Excessive fine-tuning can reduce flexibility and cause the model to become too specialized.
* **When retrieval-based approaches work better** – For PDFs, company documents, policies, or frequently updated data, RAG is usually preferred over fine-tuning.

A common rule of thumb: **If the problem can be solved with Prompt Engineering + RAG + Tools, fine-tuning may not be necessary.**


---

**Ways to Customize LLMs/SLMs with Fine-tuning**

Fine-tuning should be considered when prompts and RAG are not enough. The main approaches:

- **a. Add a Small Brain: PEFT (LoRA)**

    - **PEFT (Parameter-Efficient Fine-Tuning):** Update only a small fraction of model parameters (weights).

    - **LoRA (Low-Rank Adaptation):** Comes under PEFT, Insert small trainable matrices in layers.

    - **When to use:** Limited compute/resources; need quick adaptation.

    - **Platforms:** Hugging Face `PEFT` library, Hugging Face `AutoTrain` (low-code/no-code).

    - **Advantages:** Cost-efficient, modular.

    - **Trade-offs:** May not match full fine-tuning performance on very complex tasks.

- **b. Slightly Guided Training: SFT (Supervised Fine-Tuning)**

    - Uses labeled **input → output pairs** to guide model behavior.

    - **When to use:** Want the model to consistently follow instructions (e.g., helpdesk queries).

    - **Platforms:** Hugging Face `Transformers` library, OpenAI fine-tuning API, Anthropic fine-tuning tools.

    - **Advantages:** Strong alignment with desired outputs.

    - **Trade-offs:** Requires curated labeled data.

- **c. Full Fine-tuning**

    - Unfreeze layers and train all parameters on domain/task data.

    - **When to use:**

        - Large datasets are available.

        - You need **maximum domain specialization**.

    - **Platforms:** Hugging Face Accelerate, DeepSpeed, PyTorch Lightning.

    - **Advantages:** Best performance, fully domain-adapted.

    - **Trade-offs:** Very costly (compute + data), risk of overfitting, long training times.

---

**Best Practices for Fine-tuning**

- **Start with lightweight methods first** (LoRA/PEFT, SFT) before attempting full fine-tuning.

- **Curate high-quality data** (clean, diverse, representative of your domain).

- **Use smaller datasets wisely** — even 5k–10k good examples can outperform large noisy ones.

- **Track metrics** like accuracy, F1-score, perplexity, and run baseline vs fine-tuned comparisons.

- **Avoid catastrophic forgetting** — ensure the model still retains general reasoning ability.

- **Test for safety & bias** — fine-tuning can unintentionally introduce harmful behaviors.

- **Optimize for deployment** — consider quantization, distillation, or exporting adapters for real-world use.

- **Document the fine-tuning process** — dataset details, hyperparameters, and evaluation logs.

---

### **Parameter-Efficient Fine-Tuning (PEFT) methods**

Fine-tuning large pretrained models is often prohibitively costly due to their scale. Parameter-Efficient Fine-Tuning (PEFT) methods enable efficient adaptation of large pretrained models to various downstream applications by only fine-tuning a small number of (extra) model parameters instead of all the model's parameters. This significantly decreases the computational and storage costs. Recent state-of-the-art PEFT techniques achieve performance comparable to fully fine-tuned models.

[PEFT](https://github.com/huggingface/peft) is also a new **open-source library from Hugging Face** to enable efficient adaptation of pre-trained language models (PLMs) to various downstream applications ***without*** fine-tuning all the model's parameters.

PEFT is integrated with Transformers for easy model training and inference, and Accelerate for distributed training and inference for really big models.

PEFT currently includes techniques for:

- **LoRA:** Low-Rank Adaptation of Large Language Models
- **Prefix Tuning:** P-Tuning v2
- **P-Tuning**
- **Prompt Tuning**


### **LoRA (Low-Rank Adaptation)**

It is a technique that accelerates the fine-tuning of large models while consuming less memory.

To make fine-tuning more efficient, LoRA's approach is to represent the weight updates with two smaller matrices (called update matrices) through low-rank decomposition.

A and B are update matrices in below figure.

<br>

- These new matrices can be trained to **adapt to the new data** while keeping the overall number of changes low.
- The original weight matrix **remains frozen** and doesn't receive any further adjustments.
- To produce the final results, both the original and the adapted weights are **combined**.

### Data Preprocessing

- Create/Update the `text` column by combining the `instruction`, `input`, and `output` fields
- Shuffle the dataset
- Tokenize the dataset
- Split the dataset into training and testing sets


```python
def create_text_column(data_point):
    """Generate input text based on a task instruction, (context info.), input, and output(answer)
    :param data_point: dict: Data point
    :return: string: combined text
    """
    prefix_text = 'Below is an instruction that describes a task. Write a response that ' \
               'appropriately completes the request.\n\n'
    # Samples with additional context into.
    if data_point['input']:
        text = f"""<start_of_turn>user {prefix_text} {data_point["instruction"]} here are the inputs {data_point["input"]} <end_of_turn>\n<start_of_turn>model {data_point["output"]} <end_of_turn>"""
    # Without
    else:
        text = f"""<start_of_turn>user {prefix_text} {data_point["instruction"]} <end_of_turn>\n<start_of_turn>model {data_point["output"]} <end_of_turn>"""
    return text
```

```python
# Remove `text` field from dataset
dataset = dataset.remove_columns(['text'])

# Add the updated `text` field in the dataset
text_column = [create_text_column(data_point) for data_point in dataset]
dataset = dataset.add_column("text", text_column)
```

## Load and Quantize the Model

**Quantization** is a technique to reduce the size of deep neural networks (including LLMs) by changing the precision of the weights and biases data structure.

Quantization techniques reduce memory and computational costs by representing weights and activations with lower-precision data types like 8-bit integers (int8).

**bitsandbytes** is the easiest option for quantizing a model to 8-bit and 4-bit.

* 8-bit quantization multiplies outliers in fp16 with non-outliers in int8, converts the non-outlier values back to fp16, and then adds them together to return the weights in fp16. This reduces the degradative effect outlier values have on a model's performance.

* 4-bit quantization compresses a model even further, and it is commonly used with [QLoRA](https://hf.co/papers/2305.14314) to finetune quantized LLMs.

You can quantize a model by passing a `BitsAndBytesConfig` to `from_pretrained()` method.

To know more about bitsandbytes, refer [here](https://huggingface.co/docs/transformers/main/en/quantization/bitsandbytes?bnb=4-bit).

```python
# QLoRA config
# Create `BitsAndBytesConfig` object for Quantization

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,                            # Activate 4-bit precision base model loading
    bnb_4bit_quant_type="nf4",                    # Quantization type (fp4 or nf4)    nf4 -> Normal Float 4
    bnb_4bit_compute_dtype=torch.float16,         # Compute dtype for 4-bit base models
    bnb_4bit_use_double_quant=False,              # Whether to activate nested quantization for 4-bit base models (double quantization)
)
```

```python
# Load Model   (this code cell will take ~4-5 min to execute)

model_id = "google/gemma-2b-it"

# Load and quantize the base model
model = AutoModelForCausalLM.from_pretrained(model_id,
                                             quantization_config=bnb_config,      # note that GPU is needed for quantization
                                             device_map="auto"                    # to load the entire model on the GPU if its available
                                             )
```

### **Fine-tuning Model using LoRA**

To fine-tune a model using LoRA, you need to:

- Instantiate a base model, here it is `google/gemma-2b-it`
- Create a configuration (`LoraConfig`) where you define LoRA-specific parameters
- Pass this configuration to SFTTrainer class `peft_config = LoraConfig(...)`
- Start the training process

```python
# Create LoRA configuration object with LoRA-specific parameters

from peft import LoraConfig, get_peft_model

lora_config = LoraConfig(
    lora_alpha=32,                         # LoRA scaling factor
    r=64,                     # 8, 16, 32  # LoRA attention dimension; the rank of the update matrices
    lora_dropout=0.05,                     # Dropout probability for LoRA layers
    bias="none",                           # specifies if the bias parameters should be trained
    task_type="CAUSAL_LM",                 # telling lora that this is a Causal Language Modeling task
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj","gate_proj", "up_proj", "down_proj"]         # layers/blocks present in gemma-2b model
)

peft_model = get_peft_model(model, lora_config)
```

### Data Collator

Data collators are objects that:

- will form a batch by using a list of dataset elements as input
- may apply some processing (like padding)

One of the data collators, `DataCollatorForLanguageModeling`, can also apply some random data augmentation (like random masking) on the formed batch.

<br>

`DataCollatorForLanguageModeling` is a data collator used for language modeling. Inputs are dynamically padded to the maximum length of a batch if they are not all of the same length.

Parameters:

- ***tokenizer:*** The tokenizer used for encoding the data.
- ***mlm*** (bool, optional, default=True): Whether or not to use masked language modeling.
    - If set to False, the labels are the same as the inputs with the padding tokens ignored (by setting them to -100).
    - Otherwise, the labels are -100 for non-masked tokens and the value to predict for the masked token.
- ***return_tensors*** (str): The type of Tensor to return. Allowable values are “np”, “pt” and “tf” for numpy array, pytorch tensor, and tensorflow tensor respectively.

To know more about `DataCollatorForLanguageModeling` parameters, refer [here](https://huggingface.co/docs/transformers/v4.32.0/en/main_classes/data_collator#transformers.DataCollatorForLanguageModeling).

```python
# Create a Data collator object

import transformers

data_collator = transformers.DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm=False)
```

### Fine-tune Model *(Switch to GPU runtime)*

Fine-tune the `google/gemma-2b-it` model using the provided training arguments. Save the resulting trained model and tokenizer to a specified output directory.

The `Trainer` class provides an API for feature-complete training in PyTorch for most standard use cases.

Before instantiating your Trainer, create a `TrainingArguments` to access all the points of customization during training.

`TrainingArguments` parameters:

- ***output_dir*** (str): The output directory where the model predictions and checkpoints will be written.
- ***overwrite_output_dir*** (bool, optional, default=False): If True, overwrite the content of the output directory. Use this to continue training if output_dir points to a checkpoint directory.
- ***per_device_train_batch_size*** (int, optional, default=8): The batch size per GPU/TPU/MPS/NPU core/CPU for training.

To know more about `TrainingArguments` parameters, refer [here](https://huggingface.co/docs/transformers/v4.32.0/en/main_classes/trainer#transformers.TrainingArguments).

To know more about `Trainer` parameters, refer [here](https://huggingface.co/docs/transformers/v4.32.0/en/main_classes/trainer#transformers.Trainer).

```python
%%capture

#new code using SFTTrainer
import transformers

from trl import SFTTrainer

tokenizer.pad_token = tokenizer.eos_token
torch.cuda.empty_cache()

# Setting Hyperparamter
training_arguments = transformers.TrainingArguments(
        per_device_train_batch_size=1,         # Batch size per GPU for training
        gradient_accumulation_steps=4,         # Batch size per GPU for evaluation
        warmup_steps=10,                       # Steps for a linear warmup
        # num_train_epochs=5,                    # Number of training epochs
        max_steps=80,                         # Total number of update steps to run (overrides `num_train_epochs`)
        learning_rate=2e-4,                    # Initial learning rate (AdamW optimizer)
        logging_steps=1,                       # Log every X updates steps
        output_dir="outputs",                  # Output directory where the model predictions and checkpoints will be stored
        optim="paged_adamw_8bit",              # Optimizer to use (AdamW optimizer)
        save_strategy="steps",                 # save strategy (epoch/steps)
        save_steps=20,                         # The trainer will save a checkpoint at the end of every 20 steps
        report_to="none"                       # Disable logging to W&B (Weights and Biases)
    )


trainer = SFTTrainer(
    model=model,                               # Model to train
    train_dataset=train_data,                  # Training data
    eval_dataset=test_data,                    # Validation data
    peft_config=lora_config,                   # LoRA configuration to use
    args=training_arguments,                   # Training arguments
    data_collator=data_collator,               # Data collator to use
)
```

```python
# Start the training process
trainer.train()
```

```python
# Save the fine-tuned model in local folder

new_model = "finetuned-gemma-2b-code-instruct"      # Name of the model you will be pushing to huggingface model hub

trainer.model.save_pretrained(new_model)
```

```python
def get_completion(query: str, model, tokenizer) -> str:
    device = "cuda:0"    # cuda denotes gpu
    prompt = f"""
    <start_of_turn>user
    Below is an instruction that describes a task. Write a response that appropriately completes the request.
    {query}
    <end_of_turn>\\n<start_of_turn>model
    """

    encodeds = tokenizer(prompt, return_tensors="pt", add_special_tokens=True)
    model_inputs = encodeds.to(device)
    generated_ids = model.generate(**model_inputs, max_new_tokens=1000, do_sample=True, pad_token_id=tokenizer.eos_token_id)
    # decoded = tokenizer.batch_decode(generated_ids)          # when using batches
    decoded = tokenizer.decode(generated_ids[0], skip_special_tokens=True)
    return (decoded)
```

```python
# Inference using fine-tuned model
result = get_completion(query="implement a function to check if a number is prime in Python", model=merged_model, tokenizer=tokenizer)
print(result)
```
